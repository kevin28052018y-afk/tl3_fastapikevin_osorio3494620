```python
# pip install pandas
import pandas as pd
# import the data from a CSV file
df = pd.read_excel('data/dataset_ventas.xlsx')
# urk or path to the Excel file
#df.head()display the first 5 rows of the DataFrame
#df.info()display information about the DataFrame  
#df.tail()display the last 5 rows of the DataFrame
#df.describe()display summary statistics of the DataFrame
#df.sample()display a random sample of rows from the DataFrame
#df.shape  #display the dimensions of the DataFrame
```

```python
## Manejo de valores Faltantes y duplicados
# 1. Identificar valores faltantes en el dataset

df.isnull().sum()  #display the number of missing values in each column
```

```python
#2. eliminar filas con valores faltantes
df.dropna()  #remove rows with missing values
```

```python
# 3.rellenar valores faltantes con un valor específico
df.fillna(0)  #fill missing values with 0
```

```python
# 4. rellenar valores faltantes con la media de la columna
df['precio_unitario'] = df['precio_unitario'].fillna(df['precio_unitario'].mean())  #fill missing values in 'precio_unitario' column with the mean of that column
df.head()  #display the first 5 rows of the DataFrame to verify the changes
```

```python
#RETO1:
conteo = df['cantidad'].value_counts()
print(conteo)
#print(df.duplicated(subset=['cantidad']).sum())  #display the number of duplicate rows in the DataFrame based on 'valor_unitario' column
# 5. verificar si hay valores duplicados en el dataset
#df.duplicated().sum()  #display the number of duplicate rows in the DataFrame
#eliminar filas duplicadas
#df.drop_duplicates()  remove duplicate rows from the DataFrame
```

```python
#Reto 2:
# 1. Contar cuántas veces se repite cada número en 'cantidad'
conteo_cantidad = df['cantidad'].value_counts()
print(conteo_cantidad)

# 2. Ver específicamente cuántas veces se repite el 1
veces_uno = (df['cantidad'] == 1).sum()
print(f"El 1 se repite {veces_uno} veces")

# 3. Ver solo las cantidades que tienen duplicados (aparecen más de 1 vez)
repetidos = conteo_cantidad[conteo_cantidad > 1]
print(repetidos)
```
